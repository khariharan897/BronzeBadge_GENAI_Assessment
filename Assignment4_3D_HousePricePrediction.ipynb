## Assignment 4: 3D Visualization of House Price Regression Model

To visualize a 3D regression model for house price prediction, we'll follow these steps:
1. Generate synthetic data for 'Area (sq ft)', 'Number of Bedrooms', and 'House Price'.
2. Fit a linear regression model to this data.
3. Use `matplotlib` to create a 3D scatter plot of the data and overlay the regression plane predicted by the model.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from sklearn.linear_model import LinearRegression

# 1. Generate synthetic data for demonstration
np.random.seed(42)
num_samples = 100

# Independent variables (features)
area = np.random.normal(loc=1500, scale=300, size=num_samples) # Area in sq ft
num_bedrooms = np.random.randint(2, 6, size=num_samples)      # Number of bedrooms

# Dependent variable (target) - House Price
# Assume a linear relationship with some noise
house_price = 50 * area + 30000 * num_bedrooms + np.random.normal(loc=0, scale=50000, size=num_samples)

data = pd.DataFrame({
    'Area (sq ft)': area,
    'Number of Bedrooms': num_bedrooms,
    'House Price': house_price
})

print("Synthetic house price data generated:")
display(data.head())

# 2. Fit a linear regression model
X = data[['Area (sq ft)', 'Number of Bedrooms']]
y = data['House Price']

model = LinearRegression()
model.fit(X, y)

print(f"\nRegression Coefficients: Area={model.coef_[0]:.2f}, Bedrooms={model.coef_[1]:.2f}")
print(f"Intercept: {model.intercept_:.2f}")
```

```python
# 3. Visualize the regression model in 3D

fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Scatter plot of the actual data points
ax.scatter(data['Area (sq ft)'], data['Number of Bedrooms'], data['House Price'],
           c='blue', marker='o', label='Actual Data Points')

# Create a meshgrid for the regression plane
area_surface = np.linspace(data['Area (sq ft)'].min(), data['Area (sq ft)'].max(), 50)
bedrooms_surface = np.linspace(data['Number of Bedrooms'].min(), data['Number of Bedrooms'].max(), 50)
area_grid, bedrooms_grid = np.meshgrid(area_surface, bedrooms_surface)

# Predict house prices for the meshgrid using the trained model
price_surface = model.predict(pd.DataFrame({'Area (sq ft)': area_grid.ravel(),
                                            'Number of Bedrooms': bedrooms_grid.ravel()}))
price_surface = price_surface.reshape(area_grid.shape)

# Plot the regression plane
ax.plot_surface(area_grid, bedrooms_grid, price_surface, alpha=0.5, color='red', label='Regression Plane')

# Set labels and title
ax.set_xlabel('Area (sq ft)')
ax.set_ylabel('Number of Bedrooms')
ax.set_zlabel('House Price')
ax.set_title('3D Regression Model for House Price Prediction')

# Due to a known Matplotlib bug, legend for plot_surface needs a workaround
from matplotlib.lines import Line2D
custom_lines = [Line2D([0], [0], color='blue', lw=0, marker='o'),
                Line2D([0], [0], color='red', lw=4, alpha=0.5)]
ax.legend(custom_lines, ['Actual Data Points', 'Regression Plane'])

plt.show()
```

The 3D visualization of the house price regression model has been successfully generated! I've created synthetic data for 'Area (sq ft)', 'Number of Bedrooms', and 'House Price', then fitted a linear regression model. The plot shows the actual data points as blue spheres and the regression plane as a red surface, illustrating how the model predicts house prices based on area and the number of bedrooms.
